```python
# دیتاست فاین تیوننیگ
import os
import json
import random
import cv2
import matplotlib.pyplot as plt
import numpy as np

# =========================
# مسیر دیتاست
# =========================
DATASET = "/home/jovyan/dataset/paddleocr_dataset"

TRAIN_FILE = os.path.join(DATASET, "train.txt")
VAL_FILE   = os.path.join(DATASET, "val.txt")

# تعداد تصاویر برای بررسی
NUM_IMAGES = 10

# برای تکرارپذیری
random.seed(42)


# =========================
# خواندن train/val
# =========================
def load_samples(label_file):
    samples = []

    with open(label_file, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()

            if not line:
                continue

            image_path, annotation = line.split("\t", 1)

            try:
                annotations = json.loads(annotation)
            except json.JSONDecodeError:
                print("JSON خراب:", image_path)
                continue

            full_image_path = os.path.join(DATASET, image_path)

            if os.path.isfile(full_image_path):
                samples.append((full_image_path, annotations))

    return samples


train_samples = load_samples(TRAIN_FILE)
val_samples   = load_samples(VAL_FILE)

print("Train:", len(train_samples))
print("Val  :", len(val_samples))


# =========================
# ترکیب Train + Val
# =========================
all_samples = train_samples + val_samples

random.shuffle(all_samples)

selected = all_samples[:NUM_IMAGES]


# =========================
# رسم Polygon
# =========================
for idx, (image_path, annotations) in enumerate(selected, 1):

    image = cv2.imread(image_path)

    if image is None:
        print("تصویر خوانده نشد:", image_path)
        continue

    # BGR -> RGB
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    for ann in annotations:

        points = np.array(
            ann["points"],
            dtype=np.int32
        )

        # رسم Polygon
        cv2.polylines(
            image,
            [points],
            isClosed=True,
            color=(255, 0, 0),
            thickness=2
        )

        # نقطه‌ها
        for x, y in points:
            cv2.circle(
                image,
                (int(x), int(y)),
                3,
                (0, 255, 0),
                -1
            )

    # =========================
    # نمایش
    # =========================
    plt.figure(figsize=(12, 7))

    plt.imshow(image)

    plt.title(
        f"{idx}. {os.path.basename(image_path)} | "
        f"Boxes: {len(annotations)}"
    )

    plt.axis("off")
    plt.show()
```

```python
# توزیع موقعیت متن
import os
import json
import cv2
import numpy as np
import matplotlib.pyplot as plt

DATASET = "/home/jovyan/dataset/paddleocr_dataset"

label_files = [
    os.path.join(DATASET, "train.txt"),
    os.path.join(DATASET, "val.txt")
]

centers = []

for label_file in label_files:

    with open(label_file, "r", encoding="utf-8") as f:

        for line in f:
            line = line.strip()

            if not line:
                continue

            image_path, annotation = line.split("\t", 1)
            annotations = json.loads(annotation)

            img_path = os.path.join(DATASET, image_path)
            img = cv2.imread(img_path)

            if img is None:
                continue

            h, w = img.shape[:2]

            for ann in annotations:

                points = np.array(
                    ann["points"],
                    dtype=np.float32
                )

                # مرکز Polygon
                cx = points[:, 0].mean()
                cy = points[:, 1].mean()

                # نرمال‌سازی بین 0 و 1
                centers.append([
                    cx / w,
                    cy / h
                ])


centers = np.array(centers)

print("تعداد کل متن‌ها:", len(centers))


# =========================
# Heatmap
# =========================

plt.figure(figsize=(10, 7))

plt.hist2d(
    centers[:, 0],
    centers[:, 1],
    bins=20
)

plt.colorbar(label="Number of text boxes")

plt.xlabel("X position (0 → 1)")
plt.ylabel("Y position (0 → 1)")

plt.title("Text Position Distribution")

plt.xlim(0, 1)
plt.ylim(1, 0)

plt.grid(alpha=0.2)

plt.show()


# =========================
# آمار
# =========================

x = centers[:, 0]
y = centers[:, 1]

print("\n===== Statistics =====")

print("X mean:", round(x.mean(), 3))
print("X std :", round(x.std(), 3))

print("Y mean:", round(y.mean(), 3))
print("Y std :", round(y.std(), 3))


# =========================
# درصد نواحی
# =========================

top = (y < 0.25).mean() * 100
bottom = (y > 0.75).mean() * 100
left = (x < 0.25).mean() * 100
right = (x > 0.75).mean() * 100

center = (
    (x >= 0.25) &
    (x <= 0.75) &
    (y >= 0.25) &
    (y <= 0.75)
).mean() * 100


print("\n===== Regions =====")

print("Top 25%   :", round(top, 1), "%")
print("Bottom 25%:", round(bottom, 1), "%")
print("Left 25%  :", round(left, 1), "%")
print("Right 25% :", round(right, 1), "%")
print("Center 50%:", round(center, 1), "%")
```

```python
# فاین تیوننینگ
import os
import yaml

BASE_CONFIG = "/home/jovyan/PaddleOCR/configs/det/PP-OCRv6/PP-OCRv6_medium_det.yml"
FT_CONFIG = "/home/jovyan/PaddleOCR/configs/det/PP-OCRv6/PP-OCRv6_medium_det_finetune.yml"

DATASET = "/home/jovyan/dataset/paddleocr_dataset"
PRETRAINED = "/home/jovyan/PaddleOCR/pretrain_models/PP-OCRv6_medium_det_pretrained.pdparams"
SAVE_DIR = "/home/jovyan/PaddleOCR/output/PP-OCRv6_medium_det_finetune"

# خواندن config اصلی
with open(BASE_CONFIG, "r", encoding="utf-8") as f:
    config = yaml.safe_load(f)

# =========================
# Global
# =========================
config["Global"]["epoch_num"] = 100
config["Global"]["save_model_dir"] = SAVE_DIR
config["Global"]["save_epoch_step"] = 5
config["Global"]["eval_batch_step"] = [0, 500]
config["Global"]["pretrained_model"] = PRETRAINED

# =========================
# Optimizer
# =========================
config["Optimizer"]["lr"]["learning_rate"] = 0.0001

# =========================
# Train Dataset
# =========================
config["Train"]["dataset"]["data_dir"] = DATASET
config["Train"]["dataset"]["label_file_list"] = [
    os.path.join(DATASET, "train.txt")
]

config["Train"]["loader"]["batch_size_per_card"] = 8
config["Train"]["loader"]["num_workers"] = 8

# =========================
# Validation Dataset
# =========================
config["Eval"]["dataset"]["data_dir"] = DATASET
config["Eval"]["dataset"]["label_file_list"] = [
    os.path.join(DATASET, "val.txt")
]

config["Eval"]["loader"]["batch_size_per_card"] = 1
config["Eval"]["loader"]["num_workers"] = 4

# =========================
# ذخیره
# =========================
os.makedirs(SAVE_DIR, exist_ok=True)

with open(FT_CONFIG, "w", encoding="utf-8") as f:
    yaml.dump(
        config,
        f,
        allow_unicode=True,
        sort_keys=False
    )

print("Config ساخته شد:")
print(FT_CONFIG)

print("\nPretrained:")
print(PRETRAINED)

print("\nOutput:")
print(SAVE_DIR)
```

```python
import cv2
import matplotlib.pyplot as plt
from paddleocr import TextDetection

# =========================
# مسیرها
# =========================
MODEL_PATH = "/home/jovyan/PaddleOCR/output/PP-OCRv6_medium_det_inference"

IMAGE_PATH = "/home/jovyan/102.jpg"

# =========================
# Load model
# =========================
det_model = TextDetection(
    model_dir=MODEL_PATH,
    device="gpu",
    limit_side_len=960,
    limit_type="max",
    thresh=0.20,
    box_thresh=0.45,
    unclip_ratio=1.40,
)

# =========================
# Inference
# =========================
result = det_model.predict(IMAGE_PATH)

# =========================
# Read image
# =========================
img = cv2.imread(IMAGE_PATH)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# =========================
# Draw polygons
# =========================
count = 0

for res in result:
    data = res.get("dt_polys", [])
    scores = res.get("dt_scores", [])

    for poly, score in zip(data, scores):

        # polygon
        pts = poly.astype(int).reshape((-1, 1, 2))

        cv2.polylines(
            img_rgb,
            [pts],
            isClosed=True,
            color=(255, 0, 0),
            thickness=2
        )

        # score
        x, y = pts[0][0]

        cv2.putText(
            img_rgb,
            f"{score:.2f}",
            (x, max(y - 5, 15)),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.5,
            (255, 0, 0),
            1
        )

        count += 1

# =========================
# Show
# =========================
plt.figure(figsize=(16, 9))
plt.imshow(img_rgb)
plt.axis("off")
plt.title(f"PP-OCRv6 Medium Det | Detected: {count}")
plt.show()

print("Detected boxes:", count)
```

```python
import re
import matplotlib.pyplot as plt

# =========================
# مسیر فایل لاگ
# =========================
LOG_FILE = r"E:\step3\pp6mفاین تیون شده\train.log"


# =========================
# خواندن لاگ
# =========================
with open(LOG_FILE, "r", encoding="utf-8", errors="ignore") as f:
    log = f.read()


# =========================
# استخراج اطلاعات Training
# =========================

steps = []
losses = []
learning_rates = []

# نمونه:
# global_step: 500, epoch: 5, lr: 0.000100, loss: 6.8
pattern_train = re.compile(
    r"global_step:\s*(\d+).*?"
    r"lr:\s*([0-9.eE+-]+).*?"
    r"loss:\s*([0-9.eE+-]+)",
    re.DOTALL
)

for match in pattern_train.finditer(log):
    step = int(match.group(1))
    lr = float(match.group(2))
    loss = float(match.group(3))

    steps.append(step)
    learning_rates.append(lr)
    losses.append(loss)


# =========================
# استخراج Evaluation
# =========================

eval_steps = []
precision = []
recall = []
hmean = []

# نمونه:
# precision: 0.6595, recall: 0.4270, hmean: 0.5183

pattern_eval = re.compile(
    r"global_step:\s*(\d+).*?"
    r"precision:\s*([0-9.eE+-]+).*?"
    r"recall:\s*([0-9.eE+-]+).*?"
    r"hmean:\s*([0-9.eE+-]+)",
    re.DOTALL
)

for match in pattern_eval.finditer(log):
    step = int(match.group(1))
    p = float(match.group(2))
    r = float(match.group(3))
    h = float(match.group(4))

    eval_steps.append(step)
    precision.append(p)
    recall.append(r)
    hmean.append(h)


# =========================
# حذف داده‌های تکراری
# =========================

def remove_duplicates(x, *ys):
    data = {}

    for i, value in enumerate(x):
        data[value] = [y[i] for y in ys]

    x_sorted = sorted(data.keys())

    result = [x_sorted]

    for j in range(len(ys)):
        result.append([data[x][j] for x in x_sorted])

    return result


if steps:
    steps, losses, learning_rates = remove_duplicates(
        steps, losses, learning_rates
    )

if eval_steps:
    eval_steps, precision, recall, hmean = remove_duplicates(
        eval_steps, precision, recall, hmean
    )


# =========================
# چاپ اطلاعات
# =========================

print(f"Training points : {len(steps)}")
print(f"Evaluation points: {len(eval_steps)}")

if hmean:
    best_idx = hmean.index(max(hmean))

    print("\n===== BEST MODEL =====")
    print(f"Step      : {eval_steps[best_idx]}")
    print(f"Precision : {precision[best_idx]:.4f}")
    print(f"Recall    : {recall[best_idx]:.4f}")
    print(f"Hmean     : {hmean[best_idx]:.4f}")


# =====================================================
# نمودار 1 - Loss
# =====================================================

if steps:
    plt.figure(figsize=(10, 5))

    plt.plot(
        steps,
        losses,
        linewidth=1.5
    )

    plt.xlabel("Global Step")
    plt.ylabel("Loss")
    plt.title("PP-OCRv6 Training Loss")
    plt.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()


# =====================================================
# نمودار 2 - Learning Rate
# =====================================================

if steps:
    plt.figure(figsize=(10, 5))

    plt.plot(
        steps,
        learning_rates,
        linewidth=1.5
    )

    plt.xlabel("Global Step")
    plt.ylabel("Learning Rate")
    plt.title("PP-OCRv6 Learning Rate")
    plt.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()


# =====================================================
# نمودار 3 - Precision / Recall / Hmean
# =====================================================

if eval_steps:

    plt.figure(figsize=(10, 5))

    plt.plot(
        eval_steps,
        precision,
        marker="o",
        label="Precision"
    )

    plt.plot(
        eval_steps,
        recall,
        marker="o",
        label="Recall"
    )

    plt.plot(
        eval_steps,
        hmean,
        marker="o",
        label="Hmean"
    )

    plt.xlabel("Global Step")
    plt.ylabel("Score")
    plt.title("PP-OCRv6 Evaluation Metrics")

    plt.legend()
    plt.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()
```
